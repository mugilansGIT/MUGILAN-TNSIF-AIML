```python
import numpy as np
temperature=np.array([30, 32, 35, 28, 25, 27, 29])
print("7 days temperature:", temperature)
```

```python
print("Average temperature:",np.mean(temperature))
```

```python
print("highest temperature:",np.max(temperature))
print("lowest temperature:",np.min(temperature))
```

```python
print("Days where temperature was above 30 degrees:", np.where(temperature > 30)[0])
```

```python
print("Updated temperature array after adding 2 degrees to each day:", temperature + 2)
```
