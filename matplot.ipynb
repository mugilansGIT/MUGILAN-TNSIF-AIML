```python
import numpy as np
import matplotlib.pyplot as plt
sales=np.array([100, 150, 200, 250, 300,230])
print("6 months sales data:", sales)
```

```python
line_chart = plt.plot(sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("6 Months Sales Data")
plt.legend(["Sales"])
plt.show()
```

```python
bar_chart = plt.bar(np.arange(len(sales)), sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("6 Months Sales Data")
plt.legend(["Sales"])
plt.show()
```
