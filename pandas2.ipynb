```python
import pandas as pd
df=pd.DataFrame({"product":["A","B","C"],"price":[100,200,300],"category":["Electronics","Furniture","Clothing"],"quantity":[10,5,20]})
print("total sales of each product:\n", df["price"] * df["quantity"])
```

```python
print(df.loc[(df["price"] * df["quantity"]).idxmax()])
```

```python
print("average price of products in each category:\n", df.groupby("category")["price"].mean())
```

```python
print("quantity sold over 10:\n", df[df["quantity"] > 10])
```

```python
print(
    df.loc[(df["price"] * df["quantity"]).sort_values(ascending=False).index]
)
```
