```python
import numpy as np
import matplotlib.pyplot as plt
marks=np.array([10,36,45,52,55,60,78,86,91,100])
print("marks:",marks)
```

```python
barchart=plt.bar(np.arange(len(marks)),marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
excellent = np.sum((marks >= 80) & (marks <= 100))
good = np.sum((marks >= 50) & (marks < 80))
average = np.sum((marks >= 41) & (marks < 50))
needs_improvement=np.sum((marks >= 0) & (marks < 40))

counts = [excellent, good, average, needs_improvement]

plt.pie(
    counts,
    labels=["Excellent", "Good", "Average", "Needs Improvement"],
    autopct='%1.1f%%'
)

plt.title("Student Marks Distribution")
plt.show()
```
