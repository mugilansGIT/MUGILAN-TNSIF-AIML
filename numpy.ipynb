```python
import numpy as np
data = np.array([17,42,8,91,63,29,76,4,55,88])
print("Original marks:", data)
```

```python
print("Total marks:", np.sum(data))
```

```python
print("Average marks:", np.mean(data))
```

```python
print("Highest marks:", np.max(data))
print("Lowest marks:", np.min(data))
```

```python
print("Marks above 75:", data[data > 75])
```
